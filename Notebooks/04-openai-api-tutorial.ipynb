```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
import os
from langchain_core.messages import HumanMessage, SystemMessage

load_dotenv()
api_key=os.getenv('OPENAI_API_KEY')


# Initialize the Chat Model
llm = ChatOpenAI(
    model="gpt-4o",       # Using OpenAI's GPT-4o model
    temperature=0,        # Keeping responses deterministic
    max_tokens=None,      # No limit on response length
    timeout=None,         # No timeout restriction
    max_retries=2         # Retries in case of failure
)

# Define the conversation messages
messages = [
    ("system", "Act as an expert tax consultant that is well-versed with US tax laws. Assist user with tax questions in simple and short responses"),    
    ("human", "Simulate the impact of a proposed tax reform that adjusts the tax brackets for incomes above $100,000. How would the effective tax rates change for incomes of $90,000, $110,000, and $130,000?"),
    
]

# Send the message to the AI model
ai_msg = llm.invoke(messages)

# Print the AI’s response
print(ai_msg.content)
```
